### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

Exemplo:

```
Distância da viagem (km): 450
Consumo médio (km/l): 12.5
Preço do litro (R$): 5.89
Total de pedágios (R$): 32.40
Número de pessoas: 3

Litros necessários: 36.0
Custo do combustível: R$ 212.04
Custo total da viagem: R$ 244.44
Custo por pessoa: R$ 81.48
Tipo da distância: <class 'float'> | Tipo do número de pessoas: <class 'int'>
```

````python
def calcula_litros_necessarios(distancia: float, consumo: float):
  """Calcula a quantidade de litros de combustível necessários.
  Parâmetros:
      distancia (float): Distância da viagem em km.
      consumo (float): Consumo médio do carro em km/l.
  Retorna:
      float: Litros necessários.
  """
  return distancia / consumo

def calcula_custo_combustivel(litros: float, preco: float):
  """Calcula o custo total do combustível.
  Parâmetros:
      litros (float): Quantidade de litros de combustível.
      preco (float): Preço do litro em reais.
  Retorna:
      float: Custo do combustível.
  """
  return litros * preco

def calcula_custo_total(combustivel: float, pedagio: float):
  """Calcula o custo total da viagem (combustível mais pedágios).
  Parâmetros:
      combustivel (float): Custo do combustível.
      pedagio (float): Total gasto em pedágios.
  Retorna:
      float: Custo total.
  """
  return combustivel + pedagio

def calcula_custo_pessoa(total: float, qnt: int):
  """Calcula o custo individual por pessoa.
  Parâmetros:
      total (float): Custo total da viagem.
      qnt (int): Quantidade de pessoas.
  Retorna:
      float: Custo por pessoa.
  """
  return total / qnt

distancia = float(input("Digite a distância da viagem (km): "))
consumo = float(input("Digite o consumo médio (km/l): "))
preco = float(input("Digite o preço do litro (R$): "))
pedagio = float(input("Digite o total gasto em pedágios (R$): "))
qnt_pessoas = int(input("Digite a quantidade de pessoas: "))

litros_necessarios = calcula_litros_necessarios(distancia, consumo)
custo_combustivel = calcula_custo_combustivel(litros_necessarios, preco)
custo_total = calcula_custo_total(custo_combustivel, pedagio)
custo_pessoa = calcula_custo_pessoa(custo_total, qnt_pessoas)

print('')
print(f"Litros necessários: {litros_necessarios}")
print(f"Custo do combustível: R$ {custo_combustivel}")
print(f"Custo total da viagem: R$ {custo_total}")
print(f"Custo por pessoa: R$ {custo_pessoa}")
print(f"Tipo da distância: {type(distancia)} | Tipo do número de pessoas: {type(qnt_pessoas)}")
````

### Questão 2: Crachá do estudante (1,0 ponto)

Escreva um programa que leia o **primeiro nome** e o **sobrenome** do usuário (em duas chamadas separadas de
`input`) e monte um crachá com as informações abaixo:

1. o nome completo (primeiro nome + espaço + sobrenome);
2. o nome completo em MAIÚSCULAS e em minúsculas;
3. o total de letras do nome completo, **sem contar o espaço**;
4. as iniciais em maiúsculo, separadas por ponto (por exemplo, `A.S.`);
5. um e-mail sugerido no formato `primeironome.sobrenome@edu.br`, todo em minúsculas;
6. uma linha decorativa de `=` com o mesmo comprimento do nome completo (use o operador `*` de strings);
7. uma verificação booleana: a letra `"a"` aparece no nome completo em minúsculas? (use o operador `in` e exiba o valor `True`/`False`).

Exemplo:

```
Primeiro nome: Ana
Sobrenome: Silva

=========
Ana Silva
ANA SILVA
ana silva
Total de letras: 8
Iniciais: A.S.
E-mail: ana.silva@edu.br
Contém a letra 'a': True
=========
```

````python
def concatena_string(primeiro_nome: str, sobrenome: str):
    """Concatena o primeiro nome e o sobrenome com um espaço entre eles."""
    return primeiro_nome + " " + sobrenome

def tudo_maiusculo(nome: str):
    """Converte todos os caracteres da string para letras maiúsculas."""
    return nome.upper()

def tudo_minusculo(nome: str):
    """Converte todos os caracteres da string para letras minúsculas."""
    return nome.lower()

def calcula_total_letras(nome: str):
    """Calcula a quantidade total de letras de uma string, ignorando os espaços."""
    return len("".join(nome.split()))

def inicias_maiusculas(nome: str):
    """Retorna as iniciais de cada palavra em maiúsculo, separadas por ponto."""
    return "".join([letra[0].upper()  + "." for letra in nome.split()])

def email(primeiro_nome: str, sobrenome: str):
    """Gera um e-mail sugerido no formato primeironome.sobrenome@edu.br em minúsculas."""
    p_nome = primeiro_nome.strip().split()[0].lower()
    s_nome = sobrenome.strip().split()[-1].lower()
    return f"{p_nome}.{s_nome}@edu.br"

def contem_letra(letra: str, nome: str):
    """Verifica de forma booleana se uma determinada letra existe na string."""
    return letra in nome

def linha_decorativa(nome: str):
    """Cria uma linha decorativa com o caractere '=' com o mesmo comprimento da string fornecida."""
    return "=" * len(nome)

primeiro_nome = input("Digite o primeiro nome: ")
sobrenome = input("Digite o sobrenome: ")

nome_completo = concatena_string(primeiro_nome, sobrenome)

print("")
print(linha_decorativa(nome_completo))
print(nome_completo)
print(tudo_maiusculo(nome_completo))
print(tudo_minusculo(nome_completo))
print(f"Total de letras: {calcula_total_letras(nome_completo)}")
print(f"Iniciais: {inicias_maiusculas(nome_completo)}")
print(f"E-mail: {email(primeiro_nome, sobrenome)}")
print(f"Contém letra 'a': {contem_letra('a', tudo_minusculo(nome_completo))}")
print(linha_decorativa(nome_completo))
````

### Questão 3: Distância entre dois pontos (1,0 ponto)

Usando a instrução `import` e o módulo `math`, escreva um programa que:

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
import math

def calcula_derivada(x1: float, x2: float):
  """Calcula a diferença em valor absoluto entre duas coordenadas.
    Parâmetros:
    x1 (float): Primeira coordenada.
    x2 (float): Segunda coordenada.
    """
  return math.fabs(x1 - x2)

def calcula_distancia_pontos(dx: float, dy: float):
  """Calcula a distância euclidiana entre dois pontos dadas as diferenças dx e dy.
    Parâmetros:
    dx (float): Diferença absoluta no eixo x.
    dy (float): Diferença absoluta no eixo y.
    """
  return math.sqrt(((dx ** 2) + (dy ** 2)))

def calcula_area_circulo(raio: float):
  """Calcula a área de um círculo com base no raio fornecido.
    Parâmetros:
    raio (float): O raio do círculo.
    """
  return math.pi * (raio ** 2)

x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

dx = calcula_derivada(x1, x2)
dy = calcula_derivada(y1, y2)
distancia = (calcula_distancia_pontos(dx, dy))
area = calcula_area_circulo(distancia)

print()
print(f"dx = {round(dx, 2)}")
print(f"dy = {round(dy, 2)}")
print(f"Distância entre os pontos: {round(distancia, 2)}")
print(f"Área do círculo de raio {round(distancia, 2)}: {round(area, 2)}")
````

### Questão 7: Funções booleanas e operadores lógicos (1,0 ponto)

**(a)** Escreva uma função chamada `eh_bissexto(ano)` que **retorne** `True` se o ano for bissexto e `False` caso contrário. Use os operadores `%`, `and`, `or` e `not` em uma **única expressão booleana** no `return`

**(b)** Escreva uma função chamada `eh_triangulo(a, b, c)` que retorne `True` se três medidas podem formar um
triângulo. A condição é que cada lado seja **positivo** e que cada lado seja **menor** que a soma dos outros dois.

**(c)** Usando `for i in range(...)`, exiba para os anos de **1996 a 2005** se cada um é bissexto ou não.

**(d)** Teste `eh_triangulo` com pelo menos três combinações diferentes, exibindo o resultado de cada teste.

Exemplo:

```
1996 é bissexto? True
1997 é bissexto? False
...
2000 é bissexto? True
...
eh_triangulo(3, 4, 5) -> True
eh_triangulo(1, 2, 10) -> False
eh_triangulo(-1, 4, 5) -> False
```

````python
def eh_bissexto(ano):
  return (ano % 4 == 0) and (not (ano % 100 == 0) or (ano % 400 == 0))

def eh_triangulo(a, b, c):
  return (a < (b + c)) and (b < (a + c)) and (c < (a + b))

for i in range(1996, 2006):
  print(f"{i} é bissexto? {eh_bissexto(i)}")

print(f"eh_triangulo(3, 4, 5) -> {eh_triangulo(3, 4, 5)}")
print(f"eh_triangulo(1, 2, 10) -> {eh_triangulo(1, 2, 10)}")
print(f"eh_triangulo(-1, 4, 5) -> {eh_triangulo(-1, 4, 5)}")
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
def calcula_imc(peso: float, altura: float):
  """Calcula o Índice de Massa Corporal (IMC)."""
  return peso / (altura ** 2)

def calassificar_imc(peso: float, altura: float):
  """Valida os dados e retorna a classificação do IMC."""
  if peso <= 0 or altura <= 0:
    return None
  imc = calcula_imc(peso, altura)
  if imc < 18.5:
    return "Abaixo do peso"
  elif imc >= 18.5 and imc < 25:
    return "Peso normal"
  elif imc >= 25 and imc < 30:
    return "Sobrepeso"
  else:
    return "Obesidade"

peso = float(input("Digite o seu peso (kg): "))
altura = float(input("Digite a sua altura (metros)"))

imc = calassificar_imc(peso, altura)

if imc is None:
  print("Dados inválidos: peso e altura devem ser positivos.")
else:
  print(f"IMC: {round(calcula_imc(peso, altura), 2)} -> {imc}")
````
